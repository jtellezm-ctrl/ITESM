Carga de datos

```python
import pandas as pd
import numpy as np

archivo = "/content/1695860027103-country_vaccinations.csv"
df = pd.read_csv(archivo)

# Vista rápida
display(df.head())
print("Filas, Columnas:", df.shape)
```

a) Extraer la información

```python
print(df.columns)
display(df.sample(5))
```

b) Estructura y tipo de datos

```python
print(df.dtypes)

# Convertir fecha
df["date"] = pd.to_datetime(df["date"], errors="coerce")

print("\nTipos después de convertir date:")
print(df.dtypes)
```

c) Vacunas por Compañía

```python

df["daily_vaccinations"] = pd.to_numeric(df["daily_vaccinations"], errors="coerce")

vacunas_por_combo = (
    df.groupby("vaccines", dropna=False)["daily_vaccinations"]
      .sum(min_count=1)
      .sort_values(ascending=False)
)

display(vacunas_por_combo.head(15))
```

d) Cantidad de Vacunas aplicadas en el mundo

```python
total_mundo = df["daily_vaccinations"].sum(skipna=True)
print("Total de vacunas aplicadas en el mundo (suma daily_vaccinations):", total_mundo)
```

e) Promedio de Vacunas aplicadas por País

```python
vacunas_por_pais = df.groupby("country")["daily_vaccinations"].sum(min_count=1)
promedio_por_pais = vacunas_por_pais.mean()
print("Promedio de vacunas aplicadas por país:", promedio_por_pais)
```

f)Vacunas aplicadas en el mundo el día 29-Ene-2021

```python
fecha_objetivo = pd.to_datetime("2021-01-29")

total_290121 = df.loc[df["date"] == fecha_objetivo, "daily_vaccinations"].sum(skipna=True)
print("Total de vacunas aplicadas el 29/01/21:", total_290121)
```

g) DataFrame con Diferencias

```python
df["daily_vaccinations_raw"] = pd.to_numeric(df["daily_vaccinations_raw"], errors="coerce")

conDiferencias = df.copy()
conDiferencias["diferencias"] = (
    conDiferencias["daily_vaccinations"].fillna(0) - conDiferencias["daily_vaccinations_raw"].fillna(0)
)

display(conDiferencias[["country","date","daily_vaccinations","daily_vaccinations_raw","diferencias"]].head(10))
```

h) Período de Tiempo entre la fecha de aplicación mas antigua con la mas reciente

```python
fecha_min = df["date"].min()
fecha_max = df["date"].max()
periodo = fecha_max - fecha_min

print("Fecha mínima:", fecha_min)
print("Fecha máxima:", fecha_max)
print("Periodo en días:", periodo.days)
```

i) DataFrame con cantidad con columna CanVac

```python
conCantidad = df.copy()

# Contar vacunas separadas por coma
conCantidad["canVac"] = (
    conCantidad["vaccines"]
      .fillna("")
      .apply(lambda x: 0 if x.strip() == "" else len([v.strip() for v in x.split(",") if v.strip() != ""]))
)
display(conCantidad[["country","date","vaccines","canVac"]].head(10))
```

j) DataFrame antes del 20-Dic-2020

```python
fecha_corte = pd.to_datetime("2020-12-20")
antes20 = df[df["date"] < fecha_corte].copy()

print("Registros antes del 20/12/2020:", antes20.shape[0])
display(antes20.head())
```

k)DataFrame pfizer

```python
pfizer = df[df["vaccines"].fillna("").str.contains("Pfizer", case=False, na=False)].copy()

print("Registros con Pfizer:", pfizer.shape[0])
display(pfizer[["country","date","vaccines"]].head(10))
```

l)Guardar DataFrames en EXcel

```python
nombre_excel = "resultadosReto.xlsx"

with pd.ExcelWriter(nombre_excel, engine="openpyxl") as writer:
    conDiferencias.to_excel(writer, sheet_name="conDiferencias", index=False)
    conCantidad.to_excel(writer, sheet_name="conCantidad", index=False)
    antes20.to_excel(writer, sheet_name="antes20", index=False)
    pfizer.to_excel(writer, sheet_name="pfizer", index=False)

print("Archivo generado:", nombre_excel)
```
