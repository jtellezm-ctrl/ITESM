```python

# Reto: Viabilidad de instalación de paneles solares (Solaris)

# Constantes del problema
IRRADIACION_MEDIA = 6.36
DIAS_MES = 30
UMBRAL_VIABILIDAD = 1.20

# 1) Entradas del usuario
lado1 = float(input("Ingresa la medida del lado 1 (m): "))
lado2 = float(input("Ingresa la medida del lado 2 (m): "))
consumo_mensual = float(input("Ingresa el consumo promedio mensual del hogar (kWh): "))


eficiencia = float(input("Ingresa el porcentaje de eficiencia del panel (0 a 1): "))
while eficiencia < 0 or eficiencia > 1:
    print("La eficiencia debe estar entre 0 y 1. Ejemplo: 0.5 = 50%")
    eficiencia = float(input("Ingresa el porcentaje de eficiencia del panel (0 a 1): "))

# Validaciones
if lado1 <= 0 or lado2 <= 0:
    print("Los lados deben ser mayores que 0.")
elif consumo_mensual < 0:
    print(" El consumo mensual no puede ser negativo.")
else:
    # 2) Cálculo del área
    area = lado1 * lado2

    # 3) Producción  estimada por mes
    produccion_mes = area * IRRADIACION_MEDIA * DIAS_MES * eficiencia

    # 4) Evaluación de viabilidad
    es_viable = produccion_mes >= (UMBRAL_VIABILIDAD * consumo_mensual)

    # 5) Reporte (máximo 2 decimales)
    print("\n" + "=" * 55)
    print("REPORTE: Viabilidad de instalación de paneles solares")
    print("=" * 55)
    print(f"Área disponible (m²):                 {area:.2f}")
    print(f"Consumo mensual del hogar (kWh):      {consumo_mensual:.2f}")
    print(f"Eficiencia del panel (0 a 1):         {eficiencia:.2f}")
    print(f"Irradiación media (kWh/m²/día):       {IRRADIACION_MEDIA:.2f}")
    print(f"Días considerados por mes:            {DIAS_MES}")
    print(f"Producción estimada por mes (kWh):    {produccion_mes:.2f}")
    print("-" * 55)
    print(f"Umbral de viabilidad (120% consumo):  {(UMBRAL_VIABILIDAD * consumo_mensual):.2f} kWh")

    if es_viable:
        print("Viabilidad: VIABLE instalar paneles (cumple o supera el 120%).")
    else:
        print("Viabilidad: NO VIABLE (no alcanza el 120% del consumo).")

    print("=" * 55)
```
